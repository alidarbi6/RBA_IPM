```python
import pandas as pd
```

```python
df_1 = pd.read_excel("GOR.xlsx")
df_2 = pd.read_excel("well.xlsx")
```

```python
df = pd.DataFrame()
df["Totall_watercut"] = df_2["BSW"]/(df_2["Oil"]+df_2["BSW"])
df["PI"] = df_2["FTHP"]
df["PD"] = df_2["FLP"]
df["inlet Tem"] = df_2["UCT"]* 9/5 + 32
df["Choke"]= df_2["Choke"]/64
df["well_number"] = df_2["WellId"]
df["Date"] = df_2["Date"]
```

```python
wells = {
    well_name: well_data.copy()
    for well_name, well_data in df.groupby('well_number')
}
```

```python
last_gor = (
    df_1.dropna(subset=["GOR"])
        .groupby("WellId")["GOR"]
        .last()
)


for well_name, well_data in wells.items():
    gor_value = last_gor.get(well_name, float("nan"))
    well_data["GOR"] = gor_value

```

```python
for well_name in wells:
    wells[well_name]["Date"] = pd.to_datetime(wells[well_name]["Date"])

    wells[well_name] = wells[well_name][
        (wells[well_name]["Date"] >= "2026-05-01") &
        (wells[well_name]["Date"] < "2026-06-01")
    ].copy()
```

```python
from openserver import OpenServer
import pandas as pd

src = "C:\\Users\\abolf\\OneDrive\\Desktop\\prosper\\well.Out"
out = "C:\\Users\\abolf\\OneDrive\\Desktop\\prosper\\well_pvt.Out"

pvt_vars = {
    "Prosper.PVT.Input.Solgor": 1700,
    "Prosper.PVT.Input.Api":    42,
    "Prosper.PVT.Input.Grvgas": 0.75,
    "Prosper.PVT.Input.Watsal": 143000,
    "Prosper.PVT.Input.H2s":    0.02,
    "Prosper.PVT.Input.Co2":    0.02,
    "Prosper.PVT.Input.N2":     0,
}

def compare(c, label):
    rows = []
    for tag, want in pvt_vars.items():
        got = float(c.DoGet(tag))
        rows.append({
            "tag": tag.split(".")[-1],
            "مقدار ورودی": want,
            "مقدار در Prosper": got,
            "OK": abs(got - want) < 1e-6 * max(1, abs(want)),
        })
    df = pd.DataFrame(rows)
    print(f"\n=== {label} ===")
    print(df.to_string(index=False))
    print("همه درست‌اند" if df["OK"].all() else "اختلاف وجود دارد")
    return df

# ---------- مرحلهٔ A: نوشتن + خواندن برگشتی + ذخیره ----------
with OpenServer() as c:
    c.DoCmd('PROSPER.START()')
    try:
        c.DoCmd(f'PROSPER.OPENFILE("{src}")')
    except ValueError as e:
        print("هشدار هنگام باز کردن (رد شد):", e)

    for tag, val in pvt_vars.items():
        c.DoSet(tag, float(val))

    compare(c, "A: بعد از نوشتن")
    c.DoCmd(f'PROSPER.SAVEFILE("{out}")')
    print("فایل ذخیره شد:", out)

# ---------- مرحلهٔ B: باز کردن فایل ذخیره‌شده در اتصال جدید ----------
with OpenServer() as c:
    c.DoCmd('PROSPER.START()')
    c.DoCmd(f'PROSPER.OPENFILE("{out}")')   # اگر خطای Gas Gravity نیامد، ذخیره موفق بوده
    compare(c, "B: بعد از باز کردن دوبارهٔ فایل ذخیره‌شده")
```

```python
from openserver import OpenServer

# ---------------- تنظیمات ----------------
PATH = "C:\\Users\\abolf\\OneDrive\\Desktop\\prosper\\well_pvt.Out"

PVT = {
    "Prosper.PVT.Input.Solgor": 1700,
    "Prosper.PVT.Input.Api":    42,
    "Prosper.PVT.Input.Grvgas": 0.75,
    "Prosper.PVT.Input.Watsal": 143000,
    "Prosper.PVT.Input.H2s":    0.02,
    "Prosper.PVT.Input.Co2":    0.02,
    "Prosper.PVT.Input.N2":     0,
}

METHOD = 4   # Predict Mass Flowrate
VALVE  = 0   # ELF

# یک نمونه (واحدها: scf/STB, percent, psig, deg F, psig, inch)
SAMPLE = {
    "GOR":      1596.0,
    "WaterCut": 0.16,
    "Pin":      1191.0,
    "Tin":      204.8,
    "Pout":     866.0,
    "Choke":    0.625,
}

# برچسب خروجی‌ها به ترتیب فیلدهای Calculated Data در صفحه (حدسی؛ با Prosper تطبیق دهید)
OUT_LABELS = ["Liquid Rate (STB/d)", "Mass Flowrate (lbm/day)", "Outlet Temp (F)",
              "Critical Pressure (psig)", "Critical Rate (STB/d)", "Critical Temp (F)"]

EMPTY = 1e30   # مقدار 3.4e+35 یعنی «خالی»


def val(x):
    x = float(x)
    return None if x > EMPTY else x


# ---------------- اجرا ----------------
with OpenServer() as c:
    c.DoCmd("PROSPER.START()")
    c.DoCmd(f'PROSPER.OPENFILE("{PATH}")')

    # 1) سیال
    for tag, v in PVT.items():
        c.DoSet(tag, float(v))
    print("--- PVT در Prosper ---")
    for tag, v in PVT.items():
        got = float(c.DoGet(tag))
        print(f"{tag.split('.')[-1]:8s} ورودی={v:<10} Prosper={got:<10} {'OK' if abs(got-v) < 1e-6*max(1,abs(v)) else 'اختلاف'}")

    # 2) مدل چوک
    c.DoSet("Prosper.ANL.CHK.Method", METHOD)
    c.DoSet("Prosper.ANL.CHK.Valve", VALVE)
    print("\nMethod =", c.DoGet("Prosper.ANL.CHK.Method"),
          "| Valve =", c.DoGet("Prosper.ANL.CHK.Valve"))

    # 3) ورودی‌های نمونه (ترتیب DataNH: GOR, WC, Pin, Tin, Pout, Choke)
    order = ["GOR", "WaterCut", "Pin", "Tin", "Pout", "Choke"]
    for i, k in enumerate(order):
        c.DoSet(f"Prosper.ANL.CHK.DataNH[{i}]", SAMPLE[k])
    print("\n--- ورودی‌های چوک در Prosper ---")
    for i, k in enumerate(order):
        print(f"DataNH[{i}] {k:9s} ورودی={SAMPLE[k]:<8} Prosper={float(c.DoGet(f'Prosper.ANL.CHK.DataNH[{i}]')):.4f}")

    # 4) محاسبه (اگر خطا بدهد، چیزی نمی‌خوانیم)
    try:
        c.DoCmd("Prosper.ANL.CHK.CALC")
        ok = True
    except ValueError as e:
        ok = False
        print("\nCALC خطا داد:", e)

    # 5) نتیجه
    if ok:
        print("\n--- خروجی ---")
        for i in range(12):
            v = val(c.DoGet(f"Prosper.ANL.CHK.Data[{i}]"))
            label = OUT_LABELS[i] if i < len(OUT_LABELS) else ""
            print(f"Data[{i}] = {v}   {label}")
```

```python
from openserver import OpenServer
import pandas as pd
import numpy as np

PATH = r"C:\Users\abolf\OneDrive\Desktop\prosper\well_pvt.Out"

PVT = {
    "Prosper.PVT.Input.Solgor": 1700,
    "Prosper.PVT.Input.Api": 42,
    "Prosper.PVT.Input.Grvgas": 0.75,
    "Prosper.PVT.Input.Watsal": 143000,
    "Prosper.PVT.Input.H2s": 0.02,
    "Prosper.PVT.Input.Co2": 0.02,
    "Prosper.PVT.Input.N2": 0,
}

METHOD = 0        # Predict Mass Flowrate
CHOKE_MODEL = 4   # مدل چوک (همانی که جواب داد)

INPUT_TAGS = {
    "Pin": "Prosper.ANL.CHK.Data[0]",
    "Tin": "Prosper.ANL.CHK.Data[1]",
    "Pout": "Prosper.ANL.CHK.Data[2]",
    "Choke": "Prosper.ANL.CHK.Data[4]",
    "GOR": "Prosper.ANL.CHK.Data[16]",
    "WaterCut": "Prosper.ANL.CHK.Data[17]",
}

# نگاشت ستون‌های دیتافریم به کلیدها (اینجا را مطابق ستون‌های خودتان تنظیم کنید)
COLS = {
    "Pin": "PI",
    "Tin": "inlet Tem",
    "Pout": "PD",
    "Choke": "Choke",
    "GOR": "GOR",
    "WaterCut": "Totall_watercut",
}

# شماره Data خروجی‌ها را بعد از تطبیق با صفحه عوض کنید
OUT_TAGS = {
    "Liquid_Rate": "Prosper.ANL.CHK.Data[5]",
    "Mass_Flow": "Prosper.ANL.CHK.Data[6]",
}



def run_all(df):
    df = df.copy()
    for k in OUT_TAGS:
        df[k] = np.nan

    def start():
        c = OpenServer()
        c.connect()
        c.DoCmd("PROSPER.START()")
        c.DoCmd(f'PROSPER.OPENFILE("{PATH}")')
        for tag, v in PVT.items():
            c.DoSet(tag, float(v))
        c.DoSet("Prosper.ANL.CHK.Method", METHOD)
        c.DoSet("Prosper.ANL.CHK.Choke", CHOKE_MODEL)
        return c

    c = start()
    try:
        for idx, row in df.iterrows():
            try:
                for key, tag in INPUT_TAGS.items():
                    c.DoSet(tag, float(row[COLS[key]]))
                c.DoCmd("Prosper.ANL.CHK.CALC")
                for name, tag in OUT_TAGS.items():
                    v = float(c.DoGet(tag))
                    df.at[idx, name] = np.nan if v > 1e30 else v
            except Exception as e:
                print(f"ردیف {idx} خطا: {e}")
                try:
                    c.disconnect()
                except Exception:
                    pass
                c = start()  
    finally:
        try:
            c.disconnect()
        except Exception:
            pass
    return df

result = run_all(wells["SPH-02"])
display(result)
```
